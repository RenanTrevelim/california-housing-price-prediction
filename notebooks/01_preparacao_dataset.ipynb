# 4. Padronização das Variáveis

Os nomes das colunas foram traduzidos e padronizados para português, mantendo a convenção `snake_case`.

Essa etapa melhora a legibilidade do código e facilita a interpretação das variáveis ao longo das etapas de análise, modelagem e aplicação.

```python
dados.columns = ["longitude", "latitude", "idade_mediana_imoveis", "total_comodos", "total_quartos", "populacao",
                 "domicilios", "renda_media", "valor_media_imovel", "proximidade_oceano"]
print("Colunas Traduzidas:")
print(dados.columns.tolist())
```

# 6. Verificação de Dados Nulos e Duplicados

Nesta etapa são avaliados valores ausentes e registros duplicados, dois pontos importantes para garantir a qualidade da base antes da modelagem.

Foi identificado que apenas a variável `total_quartos` possui valores nulos, totalizando **207 registros**, o que representa aproximadamente **1% da base**. Não foram encontrados registros duplicados.

Também foi criado um resumo consolidado com tipo de dado, quantidade de nulos, percentual de nulos, valores não nulos e cardinalidade de cada variável, facilitando a inspeção geral da qualidade do dataset.

```python
print("====== Verificando Dados Nulos ======")

print(dados.isna().sum())
```

```python
print("====== Verificando Dados Duplicados ======")

print(f"\nTotal de dados duplicados -> {dados.duplicated().sum()}")
```

```python
print("====== Resumo da Base de Dados ======")

resumo = pd.DataFrame({
    "Tipo": dados.dtypes,
    "Nulos": dados.isna().sum(),
    "% Nulos": (dados.isna().mean() * 100).round(2),
    "Não nulos": dados.notna().sum(),
    "Valores Unicos": dados.nunique()
})

resumo
```

# 7. Análise dos Valores Ausentes em `total_quartos`

Como os valores nulos estão concentrados apenas na variável `total_quartos`, foi realizada uma análise específica para verificar se esses registros apresentam algum padrão relevante.

Foram comparadas médias de variáveis habitacionais, populacionais e econômicas entre registros com e sem ausência em `total_quartos`, além da distribuição desses casos por `proximidade_oceano`.

Os resultados não indicam diferenças expressivas entre os grupos, sugerindo que os valores ausentes não estão fortemente concentrados em um perfil específico da base. Por esse motivo, a imputação será tratada posteriormente dentro do pipeline de Machine Learning, evitando vazamento de dados entre treino, validação e teste.

```python
dados['total_quartos'].describe().round(3)
```

```python
df_teste = dados.copy()
df_teste["total_quartos_ausente"] = (df_teste["total_quartos"].isna())
```

```python
df_teste.groupby("total_quartos_ausente")[
    [
        "total_comodos",
        "populacao",
        "domicilios",
        "renda_media",
        "valor_media_imovel"
    ]
].mean()
```

```python
pd.crosstab(
    dados["proximidade_oceano"],
    dados["total_quartos"].isna()
)
```

# 8. Padronização da Variável `proximidade_oceano`

As categorias da variável `proximidade_oceano` foram traduzidas e padronizadas para português, mantendo uma nomenclatura consistente em `snake_case`.

Essa transformação melhora a legibilidade dos dados e facilita a interpretação das categorias durante a análise exploratória, modelagem e aplicação final.

```python
oceanos = {
    "<1H OCEAN": "menos_1h_oceano",
    "INLAND": "interior",
    "ISLAND": "ilha",
    "NEAR BAY": "proximo_baia",
    "NEAR OCEAN": "proximo_oceano"
}

dados["proximidade_oceano"] = dados["proximidade_oceano"].map(oceanos)
```

```python
dados["proximidade_oceano"].unique()
```
